# Run the pipeline

Six tasks run in order: ingest, validate, features, train, evaluate, register. The plain runner calls the same functions the Airflow dag calls. Each task can be retried. A second run of a finished pipeline leaves the alias on the same MLflow run.

From this folder, `make run` does that. This notebook reads the metrics file the runner wrote.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "06-mlops-pipelines"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 06-mlops-pipelines."
    )
for folder in (root / "lib", root / "airflow" / "dags"):
    if str(folder) not in sys.path:
        sys.path.insert(0, str(folder))
ROOT = root
print("cwd", here)
print("module", ROOT)
```
