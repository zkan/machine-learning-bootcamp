# Drift check

The monitor page compares `data/live.parquet` with `data/reference.parquet`. PSI at or above 0.2 means a human has to decide. The shipped live file matches the reference, so the table should show no column over the line.

A facilitator can replace the live file. Then you decide whether to retrain or to roll the alias back.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "06-mlops-pipelines"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 06-mlops-pipelines."
    )
for folder in (root / "lib", root / "airflow" / "dags"):
    if str(folder) not in sys.path:
        sys.path.insert(0, str(folder))
ROOT = root
print("cwd", here)
print("module", ROOT)
```
