## Quick looks with Vega-Altair

Plot a few columns before you filter. Keep the charts small. Read the axes.

```python
import altair as alt

alt.Chart(customers).mark_bar().encode(
    x=alt.X("plan:N", title="Plan"),
    y=alt.Y("count()", title="Customers"),
).properties(title="Customers by plan", width=280, height=220)
```

```python
alt.Chart(customers).mark_bar().encode(
    x=alt.X("days_since_last_order:Q", bin=alt.Bin(maxbins=30), title="Days since last order"),
    y=alt.Y("count()", title="Customers"),
    color=alt.Color("churned:N", title="Churned"),
).properties(title="Recency by churn", width=360, height=220)
```
