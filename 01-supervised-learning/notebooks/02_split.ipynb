# Split before you fit

Train rows fit the model. Validation rows choose the model and threshold. Test rows are for one final check in notebook 05.

Open `split.py` and read the two stratified splits. The row index checks for overlap between sets. It is not a feature or a customer id.


```python
import sys
from pathlib import Path

here = Path.cwd().resolve()
root = None
for candidate in [here, *here.parents]:
    if (candidate / "schema.py").is_file() and (candidate / "app.py").is_file():
        root = candidate
        break
    nested = candidate / "01-supervised-learning"
    if (nested / "schema.py").is_file() and (nested / "app.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder, from 01-supervised-learning, or from its notebooks folder."
    )
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
ROOT = root
print("cwd", here)
print("module", ROOT)
```

```python
import pandas as pd

customers = pd.read_csv(ROOT / "data" / "klongcart_customers.csv")
print(customers.shape)
customers.head()
```

```python
from frame import features_and_target
from split import split_customers

train, validation, test = split_customers(customers)
print("train", len(train), "validation", len(validation), "test", len(test))
X_train, y_train = features_and_target(train)
X_val, y_val = features_and_target(validation)
X_test, y_test = features_and_target(test)
print("churned is a label, not a feature:", "churned" not in X_train.columns)
```

```python
overlap = len(set(train.index) & set(test.index))
print("overlap train and test", overlap)
```
