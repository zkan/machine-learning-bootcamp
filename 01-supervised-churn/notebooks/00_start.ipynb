KlongCart is a fictional online grocery in Bangkok. The question today is which customers will place no paid order in the next 30 days. Yes means the team might send a small voucher. No means they leave the customer alone.

You will build that check on your own laptop. The data is synthetic. The score is for the classroom demo, not for a real shop.

## The loop

1. Start from the business question.
2. Look at the table.
3. Split rows into train, validation, and test.
4. Fit a model on the train rows.
5. Choose a threshold on the validation rows.
6. Check the test rows once.
7. Type one customer into a form and read the probability.

Work in pairs. The team owns the cost of a wasted voucher and the cost of a missed churner. Those two numbers choose the threshold.
