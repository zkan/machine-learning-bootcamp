# Reproduce one run

Fresh and Good is a fictional online grocery in Bangkok. Day 1 fit a churn model on your laptop. Today the question is whether a teammate can run that same fit.

Reproducible here means four things are versioned together.

- Code, this folder in git.
- Data, the Day 1 CSV through DVC.
- Params, `params.yaml`.
- The environment, `pyproject.toml` and `uv.lock`, on Python 3.14.10.

## This notebook

Read the pipeline files, then write the auditor note.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "05-mlops-reproducibility"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 05-mlops-reproducibility."
    )
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("cwd", here)
print("module", ROOT)

import mlflow
import pandas
import sklearn

print("python", sys.version.split()[0])
print("pandas", pandas.__version__)
print("scikit-learn", sklearn.__version__)
print("mlflow", mlflow.__version__)
```

## What DVC will run

`dvc.yaml` has three stages. `prepare` reads the Day 1 CSV. `features` writes the training table and the Feast table. `train` fits the model, logs it, and writes `data/metrics.json`.

Run `make repro` from this folder before the next cell if you have not already.


```python
from pathlib import Path
import subprocess

print((ROOT / "dvc.yaml").read_text())
print((ROOT / "params.yaml").read_text())
status = subprocess.run(
    [sys.executable, "-m", "dvc", "status"],
    cwd=ROOT,
    text=True,
    capture_output=True,
)
print(status.stdout)
print(status.stderr)
```
