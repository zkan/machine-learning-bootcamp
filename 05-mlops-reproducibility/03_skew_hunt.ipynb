# Training-serving skew hunt

Training and serving both build a row for the model. One of those builders reads a July label. The form uses the other builder, and Feast never stored the July label.

The two rows for the same customer should not match. Find the column in `lib/features.py`. Do not change it yet. Name it.

Labels are not features. July spend and `churned` are labels. A column computed from a label is still a leak.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "05-mlops-reproducibility"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 05-mlops-reproducibility."
    )
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("cwd", here)
print("module", ROOT)

import mlflow
import pandas
import sklearn

print("python", sys.version.split()[0])
print("pandas", pandas.__version__)
print("scikit-learn", sklearn.__version__)
print("mlflow", mlflow.__version__)
```

```python
from features import read_customers, serving_frame, training_frame
from schema import TARGET

customers = read_customers()
trained = training_frame(customers).drop(columns=[TARGET]).iloc[[4]].reset_index(drop=True)
served = serving_frame(customers).iloc[[4]].reset_index(drop=True)
print("same values", trained.equals(served))
print("column count", trained.shape[1])
```
