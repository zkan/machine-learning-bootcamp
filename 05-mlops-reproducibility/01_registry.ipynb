# Find the model

A file path is a bad name for the model the form should load. The registry name `fresh-and-good-churn` and the alias `champion` stay still while the version number moves.

## This notebook

Load the alias when `make repro` has already registered a version. Then pick a champion from three fake versions.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "05-mlops-reproducibility"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 05-mlops-reproducibility."
    )
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("cwd", here)
print("module", ROOT)

import mlflow
import pandas
import sklearn

print("python", sys.version.split()[0])
print("pandas", pandas.__version__)
print("scikit-learn", sklearn.__version__)
print("mlflow", mlflow.__version__)
```
