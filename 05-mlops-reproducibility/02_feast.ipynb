# Look up one customer

The form should not ask a teammate to type eight numbers. It asks for a customer id. Feast returns the feature row.

The entity is `customer`. The join key is `customer_id`. The offline store is the parquet file. The online store is a sqlite file filled by `feast materialize`.

Every row has the same event timestamp, 30 June 2026. A point-in-time request before that date has no row. A request after that date sees the snapshot. There is no earlier history. See `DATA_ASSUMPTIONS.md`.

Run `make checkpoint` once before the lookup cell.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"
os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"

here = Path.cwd().resolve()
root = None
marker = "05-mlops-reproducibility"
for candidate in [here, *here.parents]:
    if candidate.name == marker and (candidate / "lib" / "schema.py").is_file():
        root = candidate
        break
    nested = candidate / marker
    if (nested / "lib" / "schema.py").is_file():
        root = nested
        break
if root is None:
    raise FileNotFoundError(
        "Start Jupyter from the repo folder or from 05-mlops-reproducibility."
    )
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("cwd", here)
print("module", ROOT)

import mlflow
import pandas
import sklearn

print("python", sys.version.split()[0])
print("pandas", pandas.__version__)
print("scikit-learn", sklearn.__version__)
print("mlflow", mlflow.__version__)
```
