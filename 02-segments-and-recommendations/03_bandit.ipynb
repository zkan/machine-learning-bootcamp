# Four snack machines

A multi-armed bandit has a few choices and a reward each time you pick one. Epsilon-greedy picks a random machine with probability epsilon. Otherwise it picks the machine with the best average reward so far. Ties go to the lower index.

The paper game in class uses 4 machines, hidden odds, and 20 pulls per team. This repo has no cardboard machines. The cell below is that game in NumPy. Seed 42 repeats the same pulls.

The odds live in `lib/bandit.py`. Leave them unread until you have run the cell.


```python
import sys
from pathlib import Path

here = Path.cwd().resolve()
root = None
for candidate in [here, *here.parents]:
    marker = candidate / "lib" / "leaks.py"
    if marker.is_file() and (candidate / "app.py").is_file():
        root = candidate
        break
if root is None:
    raise FileNotFoundError("Start Jupyter from 02-segments-and-recommendations.")
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("module", ROOT)
```

```python
from bandit import run_epsilon_greedy

epsilon = None  # between 0 and 1
pulls = 20
seed = 42

if epsilon is None:
    raise NotImplementedError("Set epsilon.")

choices, rewards = run_epsilon_greedy(epsilon, pulls, seed)
print("choices", choices.tolist())
print("rewards", rewards.tolist())
print("mean reward", round(float(rewards.mean()), 3))
```

```python
again, _ = run_epsilon_greedy(epsilon, pulls, seed)
print("same choices", choices.tolist() == again.tolist())
```
