# Customer segments

Group customers with k-means on three June columns: days since the last order (recency), orders in 90 days (frequency), and spend in 90 days (monetary). Scale the three columns first. Do not use the July columns.

Pick k from the silhouette column. Inertia falling, then flattening, is the elbow. Use it as a check, not as a second contest.


```python
import sys
from pathlib import Path

here = Path.cwd().resolve()
root = None
for candidate in [here, *here.parents]:
    marker = candidate / "lib" / "leaks.py"
    if marker.is_file() and (candidate / "app.py").is_file():
        root = candidate
        break
if root is None:
    raise FileNotFoundError("Start Jupyter from 02-unsupervised-learning.")
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("module", ROOT)
```

```python
import pandas as pd

from leaks import DAY2_CSV
from segments import fit_labels, k_table, profile

frame = pd.read_csv(DAY2_CSV)
scores = k_table(frame)
print(scores.round(4).to_string(index=False))
```

```python
k = None  # integer with the highest silhouette

if k is None:
    raise NotImplementedError("Choose k from the table above.")
print("k", k)
```

```python
labels = fit_labels(frame, k)
summary = profile(frame, labels)
print(summary.to_string(index=False))
```

## Persona and one action

For each segment, write a short persona and one action. Use the recency, orders, spend, and churn rate in the table. An action is one sentence the shop could do this month.


```python
notes = {}
# notes[0] = ("persona", "one action")

if len(notes) != k:
    raise NotImplementedError("Write one persona and one action for every segment.")

summary["persona"] = summary["segment"].map(lambda s: notes[int(s)][0])
summary["action"] = summary["segment"].map(lambda s: notes[int(s)][1])
out = ROOT / "models" / "segments.csv"
out.parent.mkdir(exist_ok=True)
summary.to_csv(out, index=False)
print("saved", out)
```
