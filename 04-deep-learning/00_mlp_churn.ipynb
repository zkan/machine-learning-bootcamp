# Neural net on churn

A neuron takes numbers in, weighs them, adds them up, and passes the total through a function (the activation). A layer is several neurons. The loss says how wrong the prediction is. Gradient descent nudges the weights to make the loss smaller. A big nudge is a high learning rate.

The morning games are described in the README. This notebook is the same idea in scikit-learn. `MLPClassifier` is a small network. The random forest uses the Day 1 settings: 80 trees, `min_samples_leaf` 5, seed 42.

The customer file is the Day 1 file. This folder does not change it. `churned` is the label. Leave the test rows unused. The score to compare is validation ROC-AUC.


```python
import os
import sys
from pathlib import Path

os.environ["MLFLOW_ALLOW_FILE_STORE"] = "true"
os.environ["MLFLOW_DISABLE_AGENT_HINT"] = "1"

here = Path.cwd().resolve()
root = None
for candidate in [here, *here.parents]:
    for folder in (candidate, candidate / "04-deep-learning"):
        marker = folder / "lib" / "customers.py"
        if marker.is_file() and (folder / "app.py").is_file():
            root = folder
            break
    if root is not None:
        break
if root is None:
    raise FileNotFoundError("Start Jupyter from the repo folder or from 04-deep-learning.")
lib = root / "lib"
if str(lib) not in sys.path:
    sys.path.insert(0, str(lib))
ROOT = root
print("module", ROOT)
```

```python
from customers import compare_models, load_customers, split_frame

customers = load_customers()
train, validation, test = split_frame(customers)
print("rows", len(customers))
print("train", len(train), "validation", len(validation), "test", len(test))
```

## Fit both models

Exercise. `hidden` is the starter network, one layer of 32 neurons. Change the tuple and run the cell again. `(16,)` is smaller. `(64, 32)` is two layers.


```python
hidden = (32,)  # Exercise. Try (16,) or (64, 32), then run this cell again.

result = compare_models(train, validation, hidden)
print("forest validation ROC-AUC", round(result["classic_validation_roc_auc"], 3))
print("MLP validation ROC-AUC", round(result["mlp_validation_roc_auc"], 3))
print("forest train ROC-AUC", round(result["classic_train_roc_auc"], 3))
print("MLP train ROC-AUC", round(result["mlp_train_roc_auc"], 3))
print("MLP iterations", int(result["mlp"].named_steps["model"].n_iter_))
print("forest seconds", round(result["classic_seconds"], 2))
print("MLP seconds", round(result["mlp_seconds"], 2))
print("test rows left unused", len(test))
```

The MLP stops at 200 iterations. The loss has not settled. More iterations can raise the train score and lower the validation score. That is the next notebook.
